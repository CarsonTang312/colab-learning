```python
import torch

device = "cuda" if torch.cuda.is_available() else "cpu"   # 选设备
x = torch.randn(2, 4, dtype=torch.bfloat16).to(device)     # 造张量并搬进 L4 显存
w = torch.randn(4, 3, dtype=torch.bfloat16).to(device)

with torch.no_grad():                                      # 推理模式，不算梯度
    logits = x @ w                                         # 矩阵乘法
    probs = torch.softmax(logits.float(), dim=-1)          # 转概率分布
    top_p, top_idx = torch.topk(probs, k=2)                # 取前 2 名

print("形状:", probs.shape, "| 设备:", probs.device, "| Top2索引:", top_idx)
```
